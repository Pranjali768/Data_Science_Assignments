## 1.	Data Cleaning and Preparation:

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

#### ●	Load the dataset into a data frame or equivalent data structure.

```python
df = pd.read_csv('Cardiotocographic.csv')
df.head()
```

#### ●	Handle missing values appropriately (e.g., imputation, deletion).

```python
#getting dataset information
df.info()
```

```python
#checking missing values
df.isnull().sum()
```

```python
#handling missing values
df.fillna(df.mean(numeric_only = True), inplace = True)
```

```python
df.isnull().sum()
```

#### ●	Identify and correct any inconsistencies in data types (e.g., numerical values stored as strings).

```python
#chacking datatypes
print(df.dtypes)
```

#### ●	Detect and treat outliers if necessary.

```python
#checking for outliers
df.boxplot(figsize = (15,5))
plt.show()
```

```python
#outlier capping using IQR method

def outlier_cap(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 *IQR
    upper_bound = Q3 + 1.5 *IQR

    df[column] = df[column].apply(
        lambda x: lower_bound if x < lower_bound
        else upper_bound if x > upper_bound
        else x
    )

for col in df.select_dtypes(include = ['int64', 'float64']).columns:
    outlier_cap(df,col)
    
```

```python
df.boxplot(figsize = (15,5))
plt.show()
```

## 2. Statistical Summary:

##### ●	Provide a statistical summary for each variable in the dataset, including measures of central tendency (mean, median) and dispersion (standard deviation, interquartile range).

```python
# generating statitical summary
df.describe()
```

```python
#calculating Mean and Median
print('Mean:')
print(df.mean(numeric_only = True))

print('\nMedian: ')
print(df.median(numeric_only = True))
```

```python
#calculating Standard Deviation
print('Standard Deviation: ')
print(df.std(numeric_only = True))
```

```python
# calculating Interquartile Range(IQR)
Q1 = df.quantile(0.25, numeric_only =True)
Q3 = df.quantile(0.75, numeric_only= True)

IQR = Q3 - Q1

print('Interquartile Range(IQR): ')
print(IQR)
```

##### The statistical summary provides information about the central tendency and variability of the dataset. Variables with higher standard deviation and IQR show greater dispersion among observations.
