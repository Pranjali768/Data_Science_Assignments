##### 1.Create a list of 5 integers. Perform and print the result of the following operations: append, extend, insert, remove, pop, clear, index, count, sort, and reverse.

```python
#create a list
lst = [3, 4, 2, 1, 5]
print('list created is: ',lst)

#append()
lst.append(7)
print('list after append: ',lst)

#extend()
lst.extend([8,9])
print('list after append: ',lst)

#insert()
lst.insert(5,6)
print('list after insert: ',lst)

#remove()
lst.remove(8)
print('list after remove: ',lst)

#pop()
lst.pop(-1)
print('list after pop: ',lst)

#index()
print('index of 5 is: ',lst.index(5))

#count()
print('count of 4 is: ',lst.count(4))

#sort()
lst.sort()
print('sorted list: ',lst)

#reverse
lst.reverse()
print('reversedlist: ',lst)

#clear()
lst.clear()
print('after clear: ',lst)
```

##### 2. Create a tuple that stores 3 student names. Try changing the second name in the tuple. What happens? Explain why.

##### 3. Create a set of integers with some duplicate values. Print the set and explain the output.

```python
num = {30, 40, 20, 30, 50,30}

print('set: ',num)
```
